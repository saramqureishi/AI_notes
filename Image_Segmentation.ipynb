```
# Install ultralytics library
!pip install ultralytics

from ultralytics import YOLO
import cv2
from google.colab.patches import cv2_imshow

# Load pretrained YOLOv8 segmentation model directly from ultralytics hub
model = YOLO('https://github.com/ultralytics/assets/releases/download/v0.0.0/yolov8m-seg.pt')
```

```
from google.colab import drive
drive.mount('/content/drive')
path = "/content/drive/My Drive/data/"
```

```
# Read image with OpenCV
img = cv2.imread(path+'segment.jpg')
cv2_imshow(img)

# Run instance segmentation
results = model(img)

# Visualise results on image
annotated_img = results[0].plot()

# Show image in Colab
cv2_imshow(annotated_img)
```
